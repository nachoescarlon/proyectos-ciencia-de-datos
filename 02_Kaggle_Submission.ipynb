```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

df_train = pd.read_csv('train.csv')
df_test = pd.read_csv('test.csv')

def preparar_datos(df):
    datos = df.copy()
    datos['Age'] = datos['Age'].fillna(datos['Age'].median())
    datos['FamilySize'] = datos['SibSp'] + datos['Parch'] + 1
    datos['IsAlone'] = 0
    datos.loc[datos['FamilySize'] == 1, 'IsAlone'] = 1
    datos['Sex'] = datos['Sex'].map({'male': 0, 'female': 1})
    return datos

train_limpio = preparar_datos(df_train)
test_limpio = preparar_datos(df_test)

features = ['Sex', 'Age', 'FamilySize', 'IsAlone']
X_train = train_limpio[features]
y_train = train_limpio['Survived']

modelo_kaggle = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
modelo_kaggle.fit(X_train, y_train)

X_test = test_limpio[features]
predicciones_ciegas = modelo_kaggle.predict(X_test)

entrega = pd.DataFrame({
    'PassengerId': test_limpio['PassengerId'],
    'Survived': predicciones_ciegas
})

entrega.to_csv('submission_titanic.csv', index=False)
print(f"Total de pasajeros predichos: {len(entrega)}")
```
